# Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.


```python
import math

a = float(input("Nhap canh a: "))
b = float(input("Nhap canh b: "))
c = float(input("Nhap canh c: "))

if a <= 0 or b <= 0 or c <= 0 or a + b <= c or a + c <= b or b + c <= a:
    print("Khong la ba canh cua mot tam giac")
else:
    if a == b == c:
        print("Tam giac deu")
    elif a == b or a == c or b == c:
        print("Tam giac can")
    elif (abs(a**2 + b**2 - c**2) < 1e-9 or
          abs(a**2 + c**2 - b**2) < 1e-9 or
          abs(b**2 + c**2 - a**2) < 1e-9):
        print("Tam giac vuong")
    else:
        print("Tam giac thuong")

    P = a + b + c
    p = P / 2
    S = math.sqrt(p * (p-a) * (p-b) * (p-c))

    print("Chu vi:", P)
    print("Dien tich:", S)
```

# Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
if a == 0:
    if b == 0:
        if c == 0:
            print("Phuong trinh vo so nghiem")
        else:
            print("Phuong trinh vo nghiem")
    else:
        x = -c / b
        print("Nghiem x =", x)
else:
    delta = b**2 - 4*a*c

    if delta < 0:
        print("Phuong trinh vo nghiem")
    elif delta == 0:
        x = -b / (2*a)
        print("Phuong trinh co nghiem kep:", x)
    else:
        x1 = (-b + math.sqrt(delta)) / (2*a)
        x2 = (-b - math.sqrt(delta)) / (2*a)

        print("x1 =", x1)
        print("x2 =", x2)
```

# Bài 8: Giải hệ phương trình tuyến tính:
ax + by = m
cx + dy = n

```python
a = float(input("Nhap a: "))
b = float(input("Nhap b: "))
m = float(input("Nhap m: "))
c = float(input("Nhap c: "))
d = float(input("Nhap d: "))
n = float(input("Nhap n: "))

D = a*d - b*c
Dx = m*d - b*n
Dy = a*n - m*c

if D != 0:
    x = Dx / D
    y = Dy / D
    print("x =", x)
    print("y =", y)

elif Dx == 0 and Dy == 0:
    print("He phuong trinh vo so nghiem")

else:
    print("He phuong trinh vo nghiem")
```

# Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vào một 
điểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
a = float(input("Nhap toa do tam a: "))
b = float(input("Nhap toa do tam b: "))
R = float(input("Nhap ban kinh R: "))

x = float(input("Nhap toa do x: "))
y = float(input("Nhap toa do y: "))

d = (x-a)**2 + (y-b)**2

if abs(d - R**2) < 1e-9:
    print("Diem A thuoc duong tron")
else:
    print("Diem A khong thuoc duong tron")
```
